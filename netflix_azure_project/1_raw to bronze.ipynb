```python
df_raw = spark.readStream.format("cloudFiles") \
                        .option("cloudFiles.format", "csv") \
                        .option("cloudFiles.schemaLocation","abfss://silver@netflixstorageaccount.dfs.core.windows.net/checkpoint")\
                        .load("abfss://raw@netflixstorageaccount.dfs.core.windows.net")
```

```python
df_raw.writeStream.option("checkpointLocation", "abfss://silver@netflixstorageaccount.dfs.core.windows.net/checkpoint").trigger(availableNow=True).start("abfss://bronze@netflixstorageaccount.dfs.core.windows.net/netflix_titles")
```
