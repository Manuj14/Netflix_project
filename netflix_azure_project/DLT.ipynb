```python
from pyspark.sql.functions import *
from pyspark.sql.types import *
```

```python
looktables_rules= {
  "rule_1" : "showid IS NOT NULL"
}
```

```python
import dlt
@dlt.table(
    name = "gold_netflixdirectors",
)

@dlt.expect_all_or_drop(looktables_rules)

def gold_netflixdirectors():
    df = spark.readStream.format("delta").load("abfss://silver@netflixstorageaccount.dfs.core.windows.net/netflix_directors")
    return df
```

```python
@dlt.table

def gold_stg_netflixtitles():
    df = spark.read.format("delta").load("abfss://silver@netflixstorageaccount.dfs.core.windows.net/netflix_titles")
    return df
```

```python

@dlt.view

def gold_trns_netflixdirectors():
    df = spark.readStream.table("LIVE.gold_stg_netflixtitles")
    df = df.withColumn("newFlag", lit(1))
    return df
```

```python
expectations = {

    "rule1" : "newFlag is NOT NULL",
    "rule2": "showid is NOT NULL"
}
```

```python
@dlt.table(
        name = "gold_netflixtitles",
)

@dlt.expect_all_or_drop(expectations)

def gold_netflixtitles():
    df = spark.readStream.table("LIVE.gold_trns_netflixdirectors")
    return df
```
