## Parameters

```python
dbutils.widgets.text("sourcefolder", "netflix_directors")
dbutils.widgets.text("targetfolder", "netflix_directors")
```

### Variables

```python
var_src_folder = dbutils.widgets.get("sourcefolder")
var_tgt_folder = dbutils.widgets.get("targetfolder")
```

## Cast

```python
df_cast = spark.read.format("csv").option("inferSchema", True).option("header",True).load(f"abfss://bronze@netflixstorageaccount.dfs.core.windows.net/{var_src_folder}")
```

```python
display(df_cast)
```

```python
df_cast.write.mode("append").option("path",f"abfss://silver@netflixstorageaccount.dfs.core.windows.net/{var_tgt_folder}").save()
```
